## Etapa 2: Demostración de Conceptos Básicos de Python

```
print("--- ETAPA 2: Conceptos Básicos de Python ---")
```

### Paso 1: Entorno de Desarrollo y Sintaxis Básica

```
ingresos = 150000
costos = 90000
print(f"Sintaxis Básica -> Ingresos: {ingresos} | Costos: {costos}")
```

### Paso 2: Tipos de Datos Primitivos y Operaciones

```
entero_id = 101                 # int
tasa_interes = 0.15             # float
centro_costo = "Norte_Ventas"   # str
activo = True                   # bool
hoy = "2026-09-22"              # str

# Gringos: 9/11/2001, Chile: 11/09/2001, 11-09-2001. Usar el formato yyyy-mm-dd, yyyymmdd

# Operaciones matemáticas
suma_ingresos = ingresos + 25000
potencia = 5 ** 2               # 5 al cuadrado
division_entera = 10 // 3       # Devuelve 3
modulo = 10 % 3                 # Devuelve el residuo (1)

print(f"Tipos de Datos -> ID: {entero_id}, Tasa: {tasa_interes}, Centro de costo: '{centro_costo}', Estado: {activo}")
print(f"Aritmética -> Suma: {suma_ingresos}, Potencia (5^2): {potencia}, Div. Entera: {division_entera}, Módulo: {modulo}")
```

### Paso 3: Estructuras de Datos: Listas y Diccionarios

```
# Listas (Indexación base cero)
meses = ['Ene', 'Feb', 'Mar']
primer_mes = meses[0]  # Acceso a 'Ene'
print(f"Listas -> Meses cargados: {meses} | Primer elemento (índice 0): {primer_mes}")

# Diccionarios (Estructuras clave-valor)
kpi = {
    'nombre': 'EBITDA',
    'valor': 25000
}
print(f"Diccionarios -> Clave 'nombre': {kpi['nombre']} | Clave 'valor': {kpi['valor']}")
```

### Paso 4: Estructuras de Control: Sentencias Condicionales

```
margen = (ingresos - costos) / ingresos
cumple_porcentaje = 0.95

# Control de flujo con indentación estricta de Python
print("\nEjecutando lógica de condiciones (IF-ELIF-ELSE):")
if margen > 0.2:
    print(" -> Margen Saludable")
elif cumple_porcentaje >= 1.0:
    bono = 100
elif cumple_porcentaje >= 0.9:
    bono = 50
else:
    print(" -> Alerta de Margen")
    bono = 0
```

### Paso 5: Bucles e Iteraciones (WHILE y FOR)

```
print("\nEjecutando iteraciones (Bucles):")
saldo = 0
meta = 1000
ahorro = 350

# Bucle WHILE
while saldo < meta:
    saldo += ahorro
print(f" -> Bucle WHILE finalizado. Saldo final acumulado: {saldo} (Meta: {meta})")

# Bucle FOR
print(" -> Bucle FOR recorriendo lista:")
for mes in meses:
    print(f"    - Mes del año analizado: {mes}")
```

### Paso 6: Funciones

```
def calc_margen(ing, cost):
    """Calcula la tasa del margen basada en ingresos y costos"""
    return (ing - cost) / ing

margen_calculado = calc_margen(ingresos, costos)
print(f"\nFunciones -> El margen calculado por función es: {margen_calculado:.2%}")
print("-" * 60 + "\n")
```

## Etapa 4: Caso Práctico de Cierre

```
print("--- ETAPA 4: Ejecución del Caso Práctico de Cierre ---")
```

### Paso 1: Lectura de los datos de ventas

```
df_origen = pd.read_csv('ventas.csv', sep=';')
print("1. Datos originales cargados correctamente.")
```

### Paso 2: Cálculo de ganancias y márgenes

```
df_origen['Ganancia_Netas'] = df_origen['Ingresos'] - df_origen['Costos']
df_origen['Margen_Ganancia'] = (df_origen['Ingresos'] - df_origen['Costos']) / df_origen['Ingresos']
print("2. Columnas de ganancias y márgenes calculadas de forma vectorizada.")
```

### Paso 3: Filtrado de anomalías (márgenes negativos)

```
# Al usar la condición < 0, aislamos los productos que están reportando pérdidas
alertas_margen_negativo = df_origen[df_origen['Margen_Ganancia'] < 0]
print("\n3. Alertas de anomalías detectadas (Márgenes Negativos):")
print(alertas_margen_negativo[['Producto', 'Ingresos', 'Costos', 'Margen_Ganancia']])
```

### Paso 4: Exportación de alertas críticas a Excel

```
alertas_margen_negativo.to_excel('alertas_criticas_margen.xlsx', index=False)
print("\n💾 4. Caso Práctico Terminado. Reporte 'alertas_criticas_margen.xlsx' exportado con éxito.")
print("=" * 60)
```
