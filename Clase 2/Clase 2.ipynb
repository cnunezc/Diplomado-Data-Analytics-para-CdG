# CLASE 2: MANIPULACIÓN AVANZADA DE DATAFRAMES CON PANDAS Y NUMPY

```python
import numpy as np
import pandas as pd

# Desactivar advertencias para salidas mas limpias
import warnings
warnings.filterwarnings('ignore')

# Configurar formateo de Pandas para visualizaciones financieras
pd.options.display.float_format = '{:,.2f}'.format

print("="*80)
print("INICIANDO EJECUCIÓN DE EJEMPLOS PRÁCTICOS DE LA CLASE 2")
print("="*80)
```

### 1. Estructura de Datos en Pandas: Series vs DataFrame

```python
print("\n--- 1. Series vs DataFrame ---")
ventas_series = pd.Series([120000, 135000, 110000], index=['Ene', 'Feb', 'Mar'], name="Ventas_CLP")

df_gestion = pd.DataFrame({
    'Mes': ['Ene', 'Feb', 'Mar'],
    'Ventas': [120000, 135000, 110000],
    'Costos': [80000, 85000, 79000]
})

print("Series:\n", ventas_series)
print("\nDataFrame:\n", df_gestion)
```

### 2. Indexación Avanzada: `.loc` vs `.iloc`

```python
print("\n--- 2. Indexacion .loc vs .iloc ---")
# .loc busca por etiqueta/condición
feb_loc = df_gestion.loc[df_gestion['Mes'] == 'Feb', ['Ventas', 'Costos']]
# .iloc busca por posición
primeras_filas_iloc = df_gestion.iloc[0:2, 0:2]

print("Uso de .loc (Filtro 'Feb'):\n", feb_loc)
print("\nUso de .iloc (Filas 0-1, Cols 0-1):\n", primeras_filas_iloc)
```

### 3. Introducción a NumPy para Control de Gestión

```python
print("\n--- 3. Operaciones Vectorizadas con NumPy ---")
ingresos_arr = np.array([100000, 150000, 200000])
costos_arr = np.array([60000, 90000, 110000])

margen_pct = (ingresos_arr - costos_arr) / ingresos_arr
print("Margen % Vectorizado:", margen_pct)
```

### 4. Condicionales Vectorizados con `np.where()`

```python
print("\n--- 4. np.where() ---")
df_gestion['Margen_Pct'] = (df_gestion['Ventas'] - df_gestion['Costos']) / df_gestion['Ventas']
df_gestion['Status_Margen'] = np.where(df_gestion['Margen_Pct'] >= 0.30, 'OK', 'Revisar')

print(df_gestion[['Mes', 'Margen_Pct', 'Status_Margen']])
```

### 5. Reglas Multicriterio con `np.select()`

```python
print("\n--- 5. np.select() ---")
condiciones = [
    df_gestion['Ventas'] >= 130000,
    (df_gestion['Ventas'] >= 115000) & (df_gestion['Ventas'] < 130000),
    df_gestion['Ventas'] < 115000
]
valores = ['Alto', 'Medio', 'Bajo']

df_gestion['Rango_Venta'] = np.select(condiciones, valores, default='N/A')
print(df_gestion[['Mes', 'Ventas', 'Rango_Venta']])
```

### 6. Tratamiento de Valores Faltantes (NaN)

```python
print("\n--- 6. Manejo de Nulos (NaN) ---")
df_incompleto = df_gestion.copy()
df_incompleto.loc[1, 'Costos'] = np.nan

print("Cantidad de nulos por columna:\n", df_incompleto.isna().sum())
# Imputación con la media
df_incompleto['Costos'] = df_incompleto['Costos'].fillna(df_incompleto['Costos'].mean())
print("\nDataFrame Imputado:\n", df_incompleto)
```

### 7. Modificación de Tipos de Datos (Casting)

```python
print("\n--- 7. Conversión de Tipos de Datos ---")
df_tipos = pd.DataFrame({
    'Monto_Texto': ['$1.500', '$2.300', '$1.800'],
    'Fecha_Texto': ['2026-01-15', '2026-02-15', '2026-03-15']
})

df_tipos['Monto_Num'] = df_tipos['Monto_Texto'].str.replace('$', '').str.replace('.', '').astype(float)
df_tipos['Fecha_DT'] = pd.to_datetime(df_tipos['Fecha_Texto'])

print(df_tipos.dtypes)
```

### 8 & 9. Agrupaciones con `groupby()` y `.agg()`

```python
print("\n--- 8 & 9. GroupBy y Agregaciones Avanzadas ---")
np.random.seed(42)
df_operaciones = pd.DataFrame({
    'Gerencia': ['Ventas', 'Ventas', 'IT', 'IT', 'Finanzas', 'Finanzas'],
    'Centro_Costo': ['CC01', 'CC02', 'CC03', 'CC03', 'CC04', 'CC04'],
    'Gastos': [5000, 7000, 3000, 4000, 2000, 2500],
    'Presupuesto': [4500, 6500, 3500, 3800, 2100, 2400]
})

resumen = df_operaciones.groupby('Gerencia').agg(
    Gastos_Totales=('Gastos', 'sum'),
    Presupuesto_Promedio=('Presupuesto', 'mean'),
    Cantidad_Registros=('Gastos', 'count')
)
print(resumen)
```

### 10. Pivot Tables en Pandas

```python
print("\n--- 10. Pivot Table ---")
df_pivot_data = pd.DataFrame({
    'Gerencia': ['Ventas', 'Ventas', 'IT', 'IT'],
    'Mes': ['Ene', 'Feb', 'Ene', 'Feb'],
    'Gastos': [5000, 7000, 3000, 4000]
})

pivot = df_pivot_data.pivot_table(
    index='Gerencia',
    columns='Mes',
    values='Gastos',
    aggfunc='sum',
    fill_value=0,
    margins=True,
    margins_name='Total'
)
print(pivot)
```

### 11 & 12. Fusión de DataFrames (Merge y Joins)

```python
print("\n--- 11 & 12. Merge (BUSCARV / Joins) ---")
df_ventas_det = pd.DataFrame({
    'ID_Vendedor': [101, 102, 103, 104],
    'Ventas': [50000, 70000, 30000, 40000]
})

df_vendedores = pd.DataFrame({
    'ID_Vendedor': [101, 102, 103],
    'Nombre': ['Ana', 'Carlos', 'Beatriz'],
    'Zona': ['Norte', 'Centro', 'Sur']
})

df_merged = pd.merge(df_ventas_det, df_vendedores, on='ID_Vendedor', how='left')
print(df_merged)
```

### 13. Concatenación de DataFrames

```python
print("\n--- 13. Concatenación (pd.concat) ---")
df_ene = pd.DataFrame({'Mes': ['Ene'], 'Ventas': [1000]})
df_feb = pd.DataFrame({'Mes': ['Feb'], 'Ventas': [1200]})

df_consolidado_q1 = pd.concat([df_ene, df_feb], axis=0, ignore_index=True)
print(df_consolidado_q1)
```

### 14 & 15. Manejo de Fechas y Variaciones Temporales (MoM)

```python
print("\n--- 14 & 15. Fechas y Variación Porcentual (MoM) ---")
df_tiempo = pd.DataFrame({
    'Fecha': pd.date_range(start='2026-01-01', periods=4, freq='M'),
    'Ventas': [1000, 1200, 1150, 1400]
})

df_tiempo['Año'] = df_tiempo['Fecha'].dt.year
df_tiempo['Mes_Num'] = df_tiempo['Fecha'].dt.month
df_tiempo['Var_MoM_%'] = df_tiempo['Ventas'].pct_change() * 100

print(df_tiempo)
```

### 16 & 17. Métodos String (`.str`) e `.isin()`

```python
print("\n--- 16 & 17. Limpieza de Cadenas e .isin() ---")
df_cc = pd.DataFrame({
    'CC_Raw': [' cc01_admin ', 'CC02_FIN ', 'cc03_it', 'CC04_RRHH'],
    'Monto': [100, 200, 150, 300]
})

df_cc['CC_Clean'] = df_cc['CC_Raw'].str.strip().str.upper()
cc_interes = ['CC01_ADMIN', 'CC03_IT']
df_filtrado = df_cc[df_cc['CC_Clean'].isin(cc_interes)]

print("Limpio y Filtrado:\n", df_filtrado)
```

### 18. Renombrando Columnas

```python
print("\n--- 18. Renombrado ---")
df_cc.rename(columns={'CC_Clean': 'Centro_Costo', 'Monto': 'Monto_USD'}, inplace=True)
print(df_cc.columns.tolist())
```

### 19. Manejo de Duplicados

```python
print("\n--- 19. Duplicados ---")
df_dups = pd.DataFrame({
    'ID_Transaccion': [1001, 1002, 1001, 1003],
    'Monto': [500, 300, 500, 400]
})

df_sin_dups = df_dups.drop_duplicates(subset=['ID_Transaccion'], keep='last')
print(df_sin_dups)
```

### 20. Cálculo de Acumulados (YTD) con `cumsum()`

```python
print("\n--- 20. Acumulados YTD (cumsum) ---")
df_ytd = pd.DataFrame({
    'Mes': [1, 2, 3, 1, 2, 3],
    'Linea': ['Retail', 'Retail', 'Retail', 'Online', 'Online', 'Online'],
    'Ventas': [100, 150, 200, 80, 120, 160]
})

df_ytd['Ventas_YTD'] = df_ytd.groupby('Linea')['Ventas'].cumsum()
print(df_ytd)
```

### 21. Aplicación de Funciones Personalizadas con `apply()`

```python
print("\n--- 21. Uso de .apply() ---")
def calcular_comision(row):
    if row['Ventas'] > 120:
        return row['Ventas'] * 0.05
    return 0.0

df_ytd['Comision'] = df_ytd.apply(calcular_comision, axis=1)
print(df_ytd[['Linea', 'Ventas', 'Comision']])
```

### 22. Resúmenes Estadísticos Rápidos

```python
print("\n--- 22. Describe ---")
print(df_ytd.groupby('Linea')['Ventas'].describe())
```

### 23. Restablecimiento de Índices (`reset_index`)

```python
print("\n--- 23. Reset Index ---")
df_agrupado = df_ytd.groupby('Linea')['Ventas'].sum().reset_index()
print(df_agrupado)
```

### 24. Optimización de Memoria

```python
print("\n--- 24. Tipos Categoricos y Memoria ---")
df_ytd['Linea'] = df_ytd['Linea'].astype('category')
print("Memoria utilizada:\n", df_ytd.memory_usage(deep=True))
```

### 25. CASO PRÁCTICO INTEGRADOR DE CIERRE DE CLASE

```python
print("\n" + "="*80)
print("--- 25. CASO PRÁCTICO INTEGRADOR ---")
print("="*80)

# 1. Crear Maestro de Sucursales
df_maestro_sucursales = pd.DataFrame({
    'ID_Sucursal': ['SUC_01', 'SUC_02', 'SUC_03'],
    'Nombre_Sucursal': ['Central Santiago', 'Concepción', 'Antofagasta'],
    'Zona': ['Centro', 'Sur', 'Norte']
})

# 2. Crear Transacciones Mensuales
df_transacciones_2026 = pd.DataFrame({
    'ID_Sucursal': ['SUC_01', 'SUC_01', 'SUC_02', 'SUC_02', 'SUC_03', 'SUC_03'],
    'Mes': [1, 2, 1, 2, 1, 2],
    'Anio': [2026, 2026, 2026, 2026, 2026, 2026],
    'Ventas_Real': [150000, 165000, 90000, 95000, 110000, 120000],
    'Presupuesto': [140000, 140000, 95000, 95000, 100000, 100000]
})

# 3. Unir Transacciones con Maestro (Left Join)
df_caso_completo = pd.merge(
    df_transacciones_2026,
    df_maestro_sucursales,
    on='ID_Sucursal',
    how='left'
)

# 4. Ordenar y calcular YTD por Zona y Año
df_caso_completo = df_caso_completo.sort_values(['Zona', 'Anio', 'Mes'])

df_caso_completo['Ventas_YTD'] = df_caso_completo.groupby(['Zona', 'Anio'])['Ventas_Real'].cumsum()
df_caso_completo['Presupuesto_YTD'] = df_caso_completo.groupby(['Zona', 'Anio'])['Presupuesto'].cumsum()

# 5. Métricas de Cumplimiento y Desviación
df_caso_completo['Cumplimiento_%'] = (df_caso_completo['Ventas_YTD'] / df_caso_completo['Presupuesto_YTD']) * 100
df_caso_completo['Desviacion_Absoluta'] = df_caso_completo['Ventas_YTD'] - df_caso_completo['Presupuesto_YTD']
df_caso_completo['Estado'] = np.where(df_caso_completo['Cumplimiento_%'] >= 100, 'Sobre Meta', 'Bajo Meta')

print("\nREPORTE CONSOLIDADO CIERRE DE CLASE 2:")
print(df_caso_completo[[
    'Zona', 'Nombre_Sucursal', 'Mes', 'Ventas_YTD', 'Presupuesto_YTD',
    'Cumplimiento_%', 'Desviacion_Absoluta', 'Estado'
]])

print("\nEJECUCIÓN FINALIZADA CON ÉXITO.")
```
