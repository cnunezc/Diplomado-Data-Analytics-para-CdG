## 2. Tipos de Uniones en Pandas (`pd.merge()`)
Crearemos dos pequeñas tablas de prueba (Usuarios y Compras) para demostrar el funcionamiento exacto de cada tipo de Join.

```python
# Datos de prueba
usuarios_df = pd.DataFrame({
    'ID_Usuario': [1, 2, 3],
    'Nombre': ['Ana', 'Carlos', 'Elena']
})

compras_df = pd.DataFrame({
    'ID_Compra': [101, 102, 103],
    'ID_Usuario': [1, 2, 5], # El ID 5 no existe en usuarios
    'Monto': [250, 400, 150]
})

print("--- Tabla Usuarios --- ")
display(usuarios_df)
print("--- Tabla Compras ---")
display(compras_df)
```

```python
# A. Inner Join (Intersección exacta)
inner_join = pd.merge(usuarios_df, compras_df, on='ID_Usuario', how='inner')
print("1. INNER JOIN (Solo coincidencias en ambas tablas):")
display(inner_join)
```

```python
# B. Left Outer Join (Prioriza la tabla de la izquierda)
left_join = pd.merge(usuarios_df, compras_df, on='ID_Usuario', how='left')
print("2. LEFT OUTER JOIN (Todos los usuarios, con o sin compras):")
display(left_join)
```

```python
# C. Right Outer Join (Prioriza la tabla de la derecha)
right_join = pd.merge(usuarios_df, compras_df, on='ID_Usuario', how='right')
print("3. RIGHT OUTER JOIN (Todas las compras, con o sin usuario asociado):")
display(right_join)
```

```python
# D. Outer Join Completo (Full Outer Join)
outer_join = pd.merge(usuarios_df, compras_df, on='ID_Usuario', how='outer')
print(
    "4. FULL OUTER JOIN (Todos los registros de ambas tablas, rellenando con"
    " NaN):")
display(outer_join)
```

## 5. Aplicación de Uniones y Validación con el Dataset de Vuelos

Ahora, aplicaremos los diferentes tipos de `merge` a nuestros DataFrames de vuelos para ver cómo afectan la cantidad de filas y la información resultante.

### DataFrames Originales:


```python
print(f"Filas en flights_df: {flights_df.shape[0]}")
print(f"Filas en routes_df: {routes_df.shape[0]}")
print(f"Filas en aircraft_df: {aircraft_df.shape[0]}")
```

### A. INNER JOIN (Vuelos y Rutas)

Un `INNER JOIN` solo conserva las filas donde hay una coincidencia en la columna clave (`Route_ID`) en *ambos* DataFrames (`flights_df` y `routes_df`).

```python
flights_routes_inner = pd.merge(flights_df, routes_df, on='Route_ID', how='inner')
print(f"Shape de flights_df: {flights_df.shape}")
print(f"Shape de routes_df: {routes_df.shape}")
print(f"Shape de flights_routes_inner (INNER JOIN): {flights_routes_inner.shape}")
display(flights_routes_inner.head())
```

### B. LEFT JOIN (Vuelos y Rutas)

Un `LEFT JOIN` mantiene todas las filas del DataFrame de la izquierda (`flights_df`) y agrega las columnas coincidentes del DataFrame de la derecha (`routes_df`). Si no hay coincidencia, las columnas de la derecha tendrán valores `NaN`.

```python
flights_routes_left = pd.merge(flights_df, routes_df, on='Route_ID', how='left')
print(f"Shape de flights_df: {flights_df.shape}")
print(f"Shape de routes_df: {routes_df.shape}")
print(f"Shape de flights_routes_left (LEFT JOIN): {flights_routes_left.shape}")
display(flights_routes_left.head())
```

### C. RIGHT JOIN (Vuelos y Rutas)

Un `RIGHT JOIN` mantiene todas las filas del DataFrame de la derecha (`routes_df`) y agrega las columnas coincidentes del DataFrame de la izquierda (`flights_df`). Si no hay coincidencia, las columnas de la izquierda tendrán valores `NaN`.

```python
flights_routes_right = pd.merge(flights_df, routes_df, on='Route_ID', how='right')
print(f"Shape de flights_df: {flights_df.shape}")
print(f"Shape de routes_df: {routes_df.shape}")
print(f"Shape de flights_routes_right (RIGHT JOIN): {flights_routes_right.shape}")
display(flights_routes_right.head())
```

### D. OUTER JOIN (Vuelos y Rutas)

Un `OUTER JOIN` (o Full Outer Join) mantiene todas las filas de *ambos* DataFrames. Si no hay coincidencia en uno de ellos, se rellenarán con `NaN`.

```python
flights_routes_outer = pd.merge(flights_df, routes_df, on='Route_ID', how='outer')
print(f"Shape de flights_df: {flights_df.shape}")
print(f"Shape de routes_df: {routes_df.shape}")
print(f"Shape de flights_routes_outer (OUTER JOIN): {flights_routes_outer.shape}")
display(flights_routes_outer.head())
```

### Combinación de Múltiples Uniones (Flights, Routes y Aircraft)

Ahora, vamos a realizar una cadena de `INNER JOIN` para combinar la información de vuelos, rutas y aeronaves. Este es un escenario común para construir una tabla "maestra" de datos interconectados.

```python
print(f"Filas en flights_df: {flights_df.shape[0]}")
print(f"Filas en routes_df: {routes_df.shape[0]}")
print(f"Filas en aircraft_df: {aircraft_df.shape[0]}")

# Primer INNER JOIN: flights_df con routes_df
flights_routes_merged = pd.merge(flights_df, routes_df, on='Route_ID', how='inner')
print(f"Filas después de flights_df + routes_df (INNER): {flights_routes_merged.shape[0]}")

# Segundo INNER JOIN: el resultado anterior con aircraft_df
flights_routes_aircraft_merged = pd.merge(flights_routes_merged, aircraft_df, on='Aircraft_ID', how='inner')
print(f"Filas después de + aircraft_df (INNER): {flights_routes_aircraft_merged.shape[0]}")

print("\n--- Primeras filas del DataFrame combinado (Flights, Routes, Aircraft) ---")
display(flights_routes_aircraft_merged.head())
print("\n--- Información estructural del DataFrame combinado ---")
flights_routes_aircraft_merged.info()
```

### Ejemplo: Creación de Clave Compuesta

A veces, ninguna columna individual es única para identificar una fila, pero una combinación de varias columnas sí lo es. En estos casos, podemos crear una *clave compuesta*.

```python
print('\n--- Ejemplo de tabla sin clave única ---')
productos_ventas_df = pd.DataFrame({
    'ID_Producto': [101, 101, 102, 102, 103],
    'Tienda': ['A', 'A', 'B', 'A', 'C'],
    'Fecha': ['2023-01-01', '2023-01-01', '2023-01-02', '2023-01-02', '2023-01-03'],
    'Cantidad_Vendida': [10, 5, 8, 12, 7]
})
display(productos_ventas_df)
```

### Paso 1: Analizar la Unicidad de las Columnas Individuales

Primero, verificamos si alguna de las columnas existentes puede servir como clave única por sí sola. Una clave única significa que no hay valores duplicados en esa columna.

### Paso 2: Identificar la Necesidad de una Clave Compuesta

Como vemos, ninguna de las columnas (`ID_Producto`, `Tienda`, `Fecha`) es única por sí misma. Sin embargo, la combinación de `ID_Producto`, `Tienda` y `Fecha` sí identifica de manera única cada registro de venta (un producto vendido en una tienda específica en una fecha particular). Esto nos indica que necesitamos una clave compuesta.
